```python
df=df.drop_duplicates()
df
```

```python
df = df.copy()

df["rating"] = df["rating"].replace("N/A", pd.NA)

df["rating"] = df["rating"].fillna(df["rating"].mode()[0])

df["rating"]
```

```python
df = df.copy()
# df["ram"] = df["rating"].replace("N/A", pd.NA)
df["ram"] = df["ram"].fillna(df["ram"].mode()[0])
df["ram"]
```

```python
df=df.copy()
df["storage"] = df["storage"].fillna(df["storage"].mode()[0])
df["storage"]
```

```python
df=df.copy()
df["color"] = df["color"].fillna(df["color"].mode()[0])
df["color"]
```

```python
df=df.copy()
df["processor"] = df["processor"].fillna(df["processor"].mode()[0])
df["processor"]
```

```python
df = df.copy()

# Convert price to numeric
df["price"] = pd.to_numeric(
    df["price"].str.replace(",", "", regex=False),
    errors="coerce"
)

# Fill missing prices with mean
df["price"] = df["price"].fillna(df["price"].mean())

df["price"]
```

```python
df.isnull().sum()
```

```python
df.to_csv("cleaned_data.csv", index=False)
print("Data saved successfully!")
```
