```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("cleaned_data.csv")

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# 1. Price Distribution - Histogram
axes[0, 0].hist(df["price"].dropna(), bins=20)
axes[0, 0].set_title("Price Distribution")
axes[0, 0].set_xlabel("Price")
axes[0, 0].set_ylabel("Number of Products")


# 2. Average Price by Brand - Bar Chart
brand_price = df.groupby("brand")["price"].mean().sort_values(ascending=False)

axes[0, 1].bar(brand_price.index, brand_price.values)
axes[0, 1].set_title("Average Price by Brand")
axes[0, 1].set_xlabel("Brand")
axes[0, 1].set_ylabel("Average Price")
axes[0, 1].tick_params(axis="x", rotation=45)


# 3. Price vs Rating - Scatter Plot
axes[0, 2].scatter(df["price"], df["rating"])
axes[0, 2].set_title("Price vs Rating")
axes[0, 2].set_xlabel("Price")
axes[0, 2].set_ylabel("Rating")


# 4. RAM Distribution - Pie Chart
ram_count = df["ram"].value_counts()

axes[1, 0].pie(
    ram_count.values,
    labels=ram_count.index,
    autopct="%1.1f%%"
)
axes[1, 0].set_title("RAM Distribution")


# 5. Storage Distribution - Pie Chart
storage_count = df["storage"].value_counts()

axes[1, 1].pie(
    storage_count.values,
    labels=storage_count.index,
    autopct="%1.1f%%"
)
axes[1, 1].set_title("Storage Distribution")


# 6. Processor Count - Bar Chart
processor_count = df["processor"].value_counts().head(10)

axes[1, 2].bar(processor_count.index, processor_count.values)
axes[1, 2].set_title("Top Processors")
axes[1, 2].set_xlabel("Processor")
axes[1, 2].set_ylabel("Number of Products")
axes[1, 2].tick_params(axis="x", rotation=45)


plt.tight_layout()
plt.show()
```
